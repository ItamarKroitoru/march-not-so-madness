# Baseline V1

Team rating:

\[
\text{rating} = \text{wins} - \text{losses}
\]

For each historical tournament game, the higher-rated team is predicted to win. Rating ties receive half credit.

```python
from pathlib import Path
import pandas as pd

import sys

PROJECT_ROOT = Path.cwd().parents[1]
sys.path.insert(0, str(PROJECT_ROOT)) if str(PROJECT_ROOT) not in sys.path else None

from src.pipelines.baseline_v1 import run_all_seasons, summarize_results
```

```python
results = run_all_seasons()
overall = summarize_results(results)

pd.set_option("display.max_rows", 15)

display(
    results.assign(
        Accuracy=(results["Accuracy"] * 100).round(2)))
```

```python
print(f"Games: {int(overall['Games'])}")
print(f"Correct: {int(overall['Correct'])}")
print(f"Incorrect: {int(overall['Incorrect'])}")
print(f"Ties: {int(overall['Ties'])}")
print(f"Effective accuracy: {overall['Accuracy']:.2%}")
```

## Result

The first baseline achieves approximately **62.38% effective accuracy** across historical NCAA tournament games.

This serves as the reference point for future rating methods that incorporate factors such as opponent strength, margin of victory, location, and recency.
