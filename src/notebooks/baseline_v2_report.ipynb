# Context Rating V2

Team rating:

\[
\text{game value}
=
\sqrt{\text{margin}}
\times
\text{location factor}
\times
\text{opponent factor}
\]

The rating is built in two passes.

In the first pass, preliminary team ratings are calculated using margin of victory and game location.

In the second pass, each game is evaluated again using the preliminary ratings to determine whether the winner was a favorite, an approximately equal opponent, or an underdog.

For each historical tournament game, the higher-rated team is predicted to win. Rating ties receive half credit.
